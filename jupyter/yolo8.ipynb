# yolov8s

```python
import os
os.environ["MKL_SERVICE_FORCE_INTEL"] = "1"

from ultralytics import YOLO

# Load a model
# model = YOLO("my_yolo11.yaml")  # build a new model from YAML
model = YOLO("yolov8.yaml")  # load a pretrained model (recommended for training)
# model = YOLO("my_yolo11.yaml").load("weights/yolo11n.pt")  # build from YAML and transfer weights

# Train the model
results = model.train(data="mydata-linux-welds-processed.yaml", 
                      imgsz=640,  
                      device = [0,1,2,3], 
                      workers=32, 
                      cache='ram',
                      batch=64, 
                      epochs=600,
                      iou=0.4,
                      cos_lr=True,
                      warmup_epochs=5.0,
                      augment=True,
                      patience=100
                     )
```
